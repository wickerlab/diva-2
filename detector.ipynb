# Helper Functions

```python
def extract_base_dataset_name(dataname):
    if "_vs_" in dataname:
        part1 = dataname.split("_vs_")[0]
        return part1.rsplit("_", 1)[0]
    return dataname
```

```python
def prepare_benchmark_data(db_path, seed, split_target='Is_Poisoned'):
    """
    Standardized data loading, filtering, and train-test splitting 
    (mimicking the original classification benchmark approach).
    """
    df = pd.read_csv(db_path)
    df['BaseGroup'] = df['Data'].apply(extract_base_dataset_name)

    # feature_noise_svm is considered clean
    df.loc[df['Method'] == 'feature_noise_svm', 'Is_Poisoned'] = 0
    df.loc[df['Method'] == 'feature_noise_svm', 'Method'] = 'clean'
    
    drop_cols = ['Data', 'Path', 'Method', 'Rate', 'Is_Poisoned', 'error', 'BaseGroup']
    drop_cols += [c for c in df.columns if c in ['Train.Clean', 'Test.Clean', 'Train.Poison', 'Test.Poison']]
    feature_cols = [c for c in df.columns if c not in drop_cols]
    
    logger.info(f"Using all {len(feature_cols)} available features.")

    df[feature_cols] = df[feature_cols].fillna(0)

    gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=seed)
    train_idx, test_idx = next(gss.split(df[feature_cols], df[split_target], df['BaseGroup']))
    
    train_df_full = df.iloc[train_idx].copy()
    test_df = df.iloc[test_idx].copy()

    # Leave out specific poisoners as defined in classification
    poisoner_to_leave_out = ["diva_attack_xgb", "diva_attack", "diva_attack2"]
    train_df_full = train_df_full[~(train_df_full['Method'].isin(poisoner_to_leave_out))]
    test_df = test_df[~(test_df['Method'].isin(poisoner_to_leave_out))]

    # Identify methods
    poisoners = sorted([m for m in train_df_full['Method'].unique() if m != 'clean'])
    all_methods = sorted(list(train_df_full['Method'].unique()))
    
    logger.info(f"Identified {len(poisoners)} poisoners to test: {poisoners}")
    
    return train_df_full, test_df, feature_cols, poisoners, all_methods
```

```python
# Load saved detector results
save_path = os.path.join("results", "detector_checkpoint.pkl")
loaded_results = joblib.load(save_path)

loao_results = loaded_results["loao_results"]
loao_global_results = loaded_results["loao_global_results"]
all_test_preds = loaded_results["all_test_preds"]
lofo_results = loaded_results["lofo_results"]

# For downstream plots, we pool 'None' predictions from the first run (seed: 42)
combined_preds =all_test_preds[0]
preds_all_train = combined_preds[combined_preds['Holdout'] == 'None'].copy()

print(f"Results successfully loaded from {save_path}. You can now run the evaluation plot cells.")
```

# Evaluate

```python
# --- Accuracy on All Methods (Poisoners + Clean/Noise) ---
method_accs_all = []
for method in ordered_methods(all_methods):
    mask = preds_all_train['Method'] == method
    if mask.sum() > 0:
        acc = accuracy_score(preds_all_train.loc[mask, 'Is_Poisoned'], preds_all_train.loc[mask, 'Prediction'])
        method_accs_all.append({'Method': method, 'Method_Display': display_method_name(method), 'Accuracy': acc})

df_accs_all = pd.DataFrame(method_accs_all)

print("Accuracy of Model Trained on ALL Data")

custom_palette = {m: PAPER_COLOR_COMPARE for m in df_accs_all['Method_Display']}

fig_all, ax_all = plt.subplots(figsize=PAPER_FIGSIZE_WIDE)
sns.barplot(data=df_accs_all, x='Method_Display', y='Accuracy', ax=ax_all, palette=custom_palette, hue='Method_Display', legend=False, edgecolor='black', linewidth=0.6)
for patch, method in zip(ax_all.patches, df_accs_all['Method']):
    if method == 'clean':
        patch.set_hatch(METHOD_HATCHES['clean'])
ax_all.set_ylim(0, 1.1)
ax_all.set_ylabel("Accuracy")
ax_all.set_xlabel("Test Method")
ax_all.set_xticks(range(len(df_accs_all)))
ax_all.set_xticklabels(df_accs_all['Method_Display'], rotation=35, ha='right')

# Annotate exact accuracy on top of the bars
for p in ax_all.patches:
    ax_all.annotate(f"{p.get_height():.2f}", 
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center', xytext=(0, 5), 
                    textcoords='offset points', fontsize=7)

plt.tight_layout()
fig_all.savefig(os.path.join(plots_dir, "accuracy_all_train_per_method.png"), dpi=300)
```

```python
"""
Standardized horizontal bar plot containing means over 10 random seeds:
- Trained on All Data (Count)
- Trained on LOAO (Leave-One-Attack-Out)
- Trained on LOFO (Leave-One-Family-Out)
Includes black error bars representing standard deviation.
"""

actual_poisoners = [p for p in poisoners if p != "None"]

def get_stats(data_list):
    """Returns mean and standard deviation. Returns 0.0, 0.0 if empty."""
    if not data_list: return 0.0, 0.0
    return np.mean(data_list), np.std(data_list)

# Extract and calculate Mean + Std Dev from the aggregated 10-seed results
blue_stats = [get_stats(loao_results.get("None", {}).get(p, [])) for p in actual_poisoners]
blue_means, blue_stds = [s[0] for s in blue_stats], [s[1] for s in blue_stats]

red_stats = [get_stats(loao_results.get(p, {}).get(p, [])) for p in actual_poisoners]
red_means, red_stds = [s[0] for s in red_stats], [s[1] for s in red_stats]

purple_stats = [get_stats(lofo_results.get(p, [])) for p in actual_poisoners]
purple_means, purple_stds = [s[0] for s in purple_stats], [s[1] for s in purple_stats]

clean_mean, clean_std = get_stats(loao_results.get("None", {}).get("clean", []))
global_mean, global_std = get_stats(loao_global_results.get("None", []))

# Plotting Setup
fig_height = max(3.0, len(actual_poisoners) * 0.8 + 1.2)
fig, ax = plt.subplots(figsize=(7.16, fig_height))

y = np.arange(len(actual_poisoners))
height = 0.25 # Thinner bars to fit 3 in a group

# Styling for the black error bars
err_style = {'ecolor': 'black', 'elinewidth': 1.0, 'capsize': 2.5}

# 1. Plot Tripled Poisoner Bars with Error Bars
bars_blue = ax.barh(y - height, blue_means, height, xerr=blue_stds, error_kw=err_style, 
                    color=PAPER_COLOR_COUNT, edgecolor=PAPER_COLOR_INK, hatch=METHOD_HATCHES['all_data'], linewidth=0.6, label='Trained on All Data')
bars_red = ax.barh(y, red_means, height, xerr=red_stds, error_kw=err_style, 
                   color=PAPER_COLOR_COMPARE, edgecolor=PAPER_COLOR_INK, hatch=METHOD_HATCHES['loao'], linewidth=0.6, label='Trained on LOAO (Attack Zero-Shot)')
bars_purple = ax.barh(y + height, purple_means, height, xerr=purple_stds, error_kw=err_style, 
                      color=PAPER_COLOR_LOFO, edgecolor=PAPER_COLOR_INK, hatch=METHOD_HATCHES['lofo'], linewidth=0.6, label='Trained on LOFO (Family Zero-Shot)')

# 2. Add Extra Bars for Pristine (Green) and Global (Orange)
y_extra_1 = len(actual_poisoners)
y_extra_2 = len(actual_poisoners) + 1

bar_green = ax.barh(y_extra_1, clean_mean, height*1.5, xerr=clean_std, error_kw=err_style, 
                    color=PAPER_COLOR_CLEAN, edgecolor=PAPER_COLOR_INK, hatch=METHOD_HATCHES['clean'], linewidth=0.6, label='Pristine (Trained on All Data)')
bar_orange = ax.barh(y_extra_2, global_mean, height*1.5, xerr=global_std, error_kw=err_style, 
                     color=PAPER_COLOR_ACCENT, edgecolor=PAPER_COLOR_INK, hatch=METHOD_HATCHES['global'], linewidth=0.6, label='Global (Trained on All Data)')

# Set Y-axis ticks & labels
all_y = list(y) + [y_extra_1, y_extra_2]
y_labels = actual_poisoners + ['Pristine', 'Global']

ax.set_yticks(all_y)
ax.set_yticklabels(y_labels)
ax.invert_yaxis()
ax.set_xlim(0, 1.15) # Extended slightly to give room for error bars + text
ax.set_xlabel("Accuracy (Mean ± Std Dev over 10 splits)")

if len(y) > 0:
    ax.axhline(y[-1] + 0.6, color='black', linestyle='--', linewidth=0.8, alpha=0.5)
    
ax.legend(
    loc='upper center',
    bbox_to_anchor=(0.5, -0.05),
    ncol=3, 
    fontsize=7, 
    title="Evaluation Context", 
    title_fontsize=8
)

# 3. Add numerical value annotations past the end of every error bar
groups = [
    (bars_blue, blue_stds),
    (bars_red, red_stds),
    (bars_purple, purple_stds),
    (bar_green, [clean_std]),
    (bar_orange, [global_std])
]

for bars, stds in groups:
    for i, bar in enumerate(bars):
        xval = bar.get_width()
        err = stds[i]
        yval = bar.get_y() + bar.get_height() / 2
        
        # Only annotate if value is greater than 0
        if xval > 0.0:
            # Shift the text by the standard deviation so it doesn't overlap the error bar cap
            ax.text(xval + err + 0.015, yval, f"{xval:.2f}", ha='left', va='center', fontsize=7)
                
fig.tight_layout()

target_plot_path = os.path.join(plots_dir, "loao_target_centric_summary.png")
fig.savefig(target_plot_path, dpi=300, bbox_inches='tight')
print(f"Saved updated bar plot with error bars to {target_plot_path}")
```

```python
# --- Average Accuracy Across All Poisoners broken down by Rate ---
# Filter out ALL benign methods since they don't have a meaningful poisoning rate
poison_preds = preds_all_train[~preds_all_train['Method'].isin(["clean"])].copy()
print("Average Global Accuracy Across All Poisoners by Rate")

if not poison_preds.empty:
    rate_accs = []
    # Group only by Rate to get the global average across all poisoners
    for rate, group in poison_preds.groupby('Rate'):
        acc = accuracy_score(group['Is_Poisoned'], group['Prediction'])
        n = len(group)
        ci = 1.96 * np.sqrt(acc * (1 - acc) / n) if n > 0 else 0
        rate_accs.append({'Rate': rate, 'Accuracy': acc, 'CI_Lower': max(0, acc - ci), 'CI_Upper': min(1, acc + ci)})
        
    df_rate_accs = pd.DataFrame(rate_accs)
    df_rate_accs['Rate'] = pd.to_numeric(df_rate_accs['Rate'])
    df_rate_accs = df_rate_accs.sort_values(by='Rate')
    
    # Plot average accuracy with a 95% binomial confidence interval
    fig_rate, ax_rate = plt.subplots(figsize=PAPER_FIGSIZE_COL)
    ax_rate.plot(df_rate_accs['Rate'], df_rate_accs['Accuracy'], marker='o', 
                 linewidth=1.4, markersize=3.5, color=PAPER_COLOR_COMPARE)
    ax_rate.fill_between(df_rate_accs['Rate'], df_rate_accs['CI_Lower'], df_rate_accs['CI_Upper'],
                         color=PAPER_COLOR_COMPARE, alpha=0.22, linewidth=0)
    
    ax_rate.set_ylim(0, 1.05)
    ax_rate.set_ylabel("Average Accuracy")
    ax_rate.set_xlabel("Poisoning Rate")
    
    plt.grid(True, linestyle='--', alpha=0.55)
    plt.tight_layout()
    
    fig_rate.savefig(os.path.join(plots_dir, "accuracy_all_train_by_rate_avg.png"), dpi=300)
```

```python
# =========================================================================
# Global vs. Clean Accuracy per Threshold Plot (0.05 Steps)
# =========================================================================
thresholds_to_test = np.arange(0.0, 1.05, 0.05)
global_accs = []
clean_accs_avg = []
global_ci_lower = []
global_ci_upper = []
clean_ci_lower = []
clean_ci_upper = []

for t in thresholds_to_test:
    # Apply current threshold (t)
    t_preds = (preds_all_train['Prediction_Prob'] >= t).astype(int)
    
    # 1. Global Accuracy
    global_acc = accuracy_score(preds_all_train['Is_Poisoned'], t_preds)
    global_accs.append(global_acc)
    global_n = len(preds_all_train)
    global_ci = 1.96 * np.sqrt(global_acc * (1 - global_acc) / global_n) if global_n > 0 else 0
    global_ci_lower.append(max(0, global_acc - global_ci))
    global_ci_upper.append(min(1, global_acc + global_ci))
    
    # 2. Clean Accuracy (Now naturally includes feature_noise_svm)
    mask_clean = preds_all_train['Method'] == 'clean'
    
    acc_c = accuracy_score(
        preds_all_train.loc[mask_clean, 'Is_Poisoned'], 
        t_preds[mask_clean]
    ) if mask_clean.sum() > 0 else 0
    
    clean_accs_avg.append(acc_c)
    clean_n = mask_clean.sum()
    clean_ci = 1.96 * np.sqrt(acc_c * (1 - acc_c) / clean_n) if clean_n > 0 else 0
    clean_ci_lower.append(max(0, acc_c - clean_ci))
    clean_ci_upper.append(min(1, acc_c + clean_ci))

# Generate the Plot
fig_acc_thresh, ax_acc_thresh = plt.subplots(figsize=PAPER_FIGSIZE_COL)

ax_acc_thresh.plot(thresholds_to_test, global_accs, 
                    label='Global Accuracy', color=PAPER_COLOR_ACCENT, linewidth=1.4, marker='o', markersize=3)
ax_acc_thresh.fill_between(thresholds_to_test, global_ci_lower, global_ci_upper,
                           color=PAPER_COLOR_ACCENT, alpha=0.20, linewidth=0)
ax_acc_thresh.plot(thresholds_to_test, clean_accs_avg, 
                    label='Clean Accuracy', color=PAPER_COLOR_CLEAN, linewidth=1.4, marker='s', markersize=3, linestyle='--')
ax_acc_thresh.fill_between(thresholds_to_test, clean_ci_lower, clean_ci_upper,
                           color=PAPER_COLOR_CLEAN, alpha=0.20, linewidth=0)

ax_acc_thresh.set_title(f"Accuracy vs. Decision Threshold")
ax_acc_thresh.set_xlabel("Decision Threshold")
ax_acc_thresh.set_ylabel("Accuracy")
ax_acc_thresh.set_xticks(np.arange(0.0, 1.05, 0.10))
ax_acc_thresh.set_xticklabels([f"{x:.1f}" for x in np.arange(0.0, 1.05, 0.10)])
ax_acc_thresh.set_xlim(0, 1)
ax_acc_thresh.set_ylim(0, 1.05)

ax_acc_thresh.legend(loc='lower right', frameon=True)
ax_acc_thresh.grid(True, linestyle='--', alpha=0.55)

plt.tight_layout()
acc_thresh_path = os.path.join(plots_dir, "accuracy_vs_threshold_steps.png")
fig_acc_thresh.savefig(acc_thresh_path, dpi=300)
```
